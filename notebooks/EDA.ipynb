```python
data = pd.read_csv("../data/yahooFinanceData.csv")

def data_cleaning(df):

    print("----------------Data Cleaning----------------")

    print(f"Original data shape: {df.shape}")

    cleaned_data = df.copy()

    #Remove duplicates
    dups = cleaned_data.duplicated().sum()
    cleaned_data = cleaned_data.drop_duplicates()
    print(f"Removed {dups} duplicates")

    #remove missing values
    n_missing = cleaned_data.isnull().sum().sum()
    cleaned_data = cleaned_data.dropna()
    print(f"Removed {n_missing} values")

    #convert sting into numeric values
    numeric_columns = ['Open', 'High', 'Low', 'Close', 'Volume']
    for col in numeric_columns:
        cleaned_data[col] = pd.to_numeric(cleaned_data[col], errors='coerce')

    #remove rows with invalid values
    invalid = (cleaned_data['Open'] <= 0) | (cleaned_data['High'] <= 0) | (cleaned_data['Low'] <= 0) | (cleaned_data['Close'] <= 0)
    cleaned_data = cleaned_data[~invalid]
    print(f"Removed {invalid.sum()} rows with invalid values")

    #logical inconsistencies
    inconsistent = (cleaned_data['High'] < cleaned_data['Low']) | (cleaned_data['High'] < cleaned_data['Close']) | (cleaned_data['Low'] > cleaned_data['Close'])
    cleaned_data = cleaned_data[~inconsistent]
    print(f"Removed {inconsistent.sum()} rows with logical inconsistencies")

    cleaned_data = cleaned_data.sort_index()

    

    print("Cleaned data shape: ", cleaned_data.shape)

    print("Cleaned data completed")

    print("----------------Data Cleaning Complete----------------")
    return cleaned_data


cleaned = data_cleaning(data)
cleaned

data.to_csv("../data/yahooFinanceDataCleaned.csv")
```

```python
from sklearn.preprocessing import StandardScaler
import pandas as pd
import numpy as np

data = pd.read_csv("../data/yahooFinanceDataCleaned.csv", header=[0,1], index_col=0)

print("original data statistics:")
print(data.describe())

data.columns = data.columns.get_level_values(0) 

features_to_scale = ['Open', 'High', 'Low', 'Close', 'Volume']

scaler = StandardScaler()
data_normalized = data.copy()
data_normalized[features_to_scale] = scaler.fit_transform(data[features_to_scale])

data_normalized.to_csv("../data/normalized_data.csv")

print("original data statistics:")
print(data[features_to_scale].describe())

print("\nNormalized data statistics:")
print(data_normalized[features_to_scale].describe())

```
